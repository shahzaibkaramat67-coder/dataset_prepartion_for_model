```python
for col in X_train.columns:
    Q1 = X_train[col].quantile(0.25)
    Q3 = X_train[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outlier_mask = ((X_train[col] < lower) | (X_train[col] > upper))
    outlier_class = Y_train[outlier_mask]
    print(f"column {col}")
    print(f"Number of Outlier :", outlier_mask.sum())
    print(outlier_class.value_counts(normalize=True))
```
