```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
numaricle_colums = X.select_dtypes(include="number").columns
categorical_colums = X.select_dtypes(include="object").columns
preprocessor = ColumnTransformer(
    transformers=[("num", StandardScaler(), numaricle_colums),
                  ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_colums)]
)
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.2, random_state=42, stratify=Y)
X_train.shape, X_test.shape, Y_train.shape, Y_test.shape
encoding_mini_llm = preprocessor.fit(X_train)
encoding_mini_llm
```

```python
X_train = preprocessor.transform(X_train)
X_test = preprocessor.transform(X_test)
X_train.shape, X_test.shape
```

```python

from sklearn.feature_selection import SelectKBest, f_classif

selector = SelectKBest(
    score_func=f_classif,
    k="all"
)

X_train_selecetde = selector.fit_transform(X_train, Y_train)
X_test_selected = selector.transform(X_test)
selector
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import GridSearchCV
# with 20 features
model = RandomForestClassifier(
    class_weight="balanced",
    n_estimators=50,
    random_state=42
)
model.fit(X_train, Y_train)
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_pred, Y_test)
print(classification_report(y_pred, Y_test))
cm = confusion_matrix(y_pred, Y_test)
print("cm :", cm)

score = cross_val_score(
    model,
    X_train,
    Y_train,
    cv=5,
    scoring="accuracy"
)

print("score :", score)
print("score mean :", score.mean())

params = {
    "n_estimators" : [100, 150],
    "max_depth" : [None, 10, 20]
}

Grid = GridSearchCV(model, params, cv=5, scoring="accuracy")

Grid.fit(X_train, Y_train)

print("grid best params :",Grid.best_params_)
print("grid best score :",Grid.best_score_)


base_model = Grid.best_estimator_
y_pred = base_model.predict(X_test)

print("classification report :", classification_report(Y_test, y_pred))

```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import GridSearchCV
model = DecisionTreeClassifier(
    random_state=42
)
model.fit(X_train, Y_train)
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_pred, Y_test)
print("accuracy :", accuracy)
print(classification_report(y_pred, Y_test))
cm = confusion_matrix(y_pred, Y_test)
print("cm :", cm)

score = cross_val_score(
    model,
    X_train,
    Y_train,
    cv=5,
    scoring="accuracy"
)

print("score :", score)
print("score mean :", score.mean())

param_grid = {
    "max_depth" : [None, 15, 20],
    "min_samples_split" : [2, 5, 10, 20],
    "min_samples_leaf" : [1, 2, 5, 10],
    "criterion" : ["gini", "entropy"]
}

Grid = GridSearchCV(model, param_grid, cv=5, scoring="f1", n_jobs= -1)

Grid.fit(X_train, Y_train)

print("grid best params :",Grid.best_params_)
print("grid best score :",Grid.best_score_)


base_model = Grid.best_estimator_
y_pred = base_model.predict(X_test)

print("classification report :", classification_report(Y_test, y_pred))

```
